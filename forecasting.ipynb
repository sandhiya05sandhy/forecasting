```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
a=pd.read_csv("/content/daily_dataset.csv")
```

```python
a
```

```python
a["energy_count"]=pd.to_datetime(a["energy_count"])
a["energy_count"]
```

```python
a=a.sort_values("energy_count")
a
```

```python
data=a["energy_count"]
data
```

```python
from statsmodels.tsa.arima.model import ARIMA
model=ARIMA(data,order=(5,1,0))
model_fit=model.fit()
forecast = model_fit.forecast(steps=10)
```

```python
print("Next 10 days energy max: ")
print("DAY 1:",forecast.iloc[0])
print("DAY 2:",forecast.iloc[1])
print("DAY 3:",forecast.iloc[2])
print("DAY 4:",forecast.iloc[3])
print("DAY 5:",forecast.iloc[4])
print("DAY 6:",forecast.iloc[5])
print("DAY 7:",forecast.iloc[6])
print("DAY 8:",forecast.iloc[7])
print("DAY 9:",forecast.iloc[8])
print("DAY 10:",forecast.iloc[9])
```
